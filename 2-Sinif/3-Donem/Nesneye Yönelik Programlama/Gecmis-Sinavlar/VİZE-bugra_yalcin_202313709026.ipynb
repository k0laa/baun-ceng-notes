**Soru 1:** Bir Calisan sınıfı oluşturun. Bu sınıfın her çalışan için ad, soyad ve maaş bilgilerini tutmasını sağlayın. Ayrıca, maaşı yüzde belirli bir oranda artırmak için maas_artir adında bir fonksiyon oluşturun.

Cevap:

```python
class Employee:
    def __init__(self, name: str, surname: str, salary: float):
        self.name = name
        self.surname = surname
        self.salary = salary

    def increase_salary(self, percentage: float):
        self.salary += self.salary * percentage / 100
        print(f"{self.name} {self.surname} maaşı güncellendi: {self.salary} TL")
```

**Soru 2:**  Bir alışveriş sepeti uygulaması için Urun ve Sepet adında iki sınıf tanımlayın. Urun sınıfı, ürün adı ve fiyatını içermelidir.
Sepet sınıfı ise, sepete ürün eklemeli ve toplam tutarı hesaplamalıdır.

Cevap:

```python
class Product:
    def __init__(self, name: str, price: float):
        self.name = name
        self.price = price
```

```python
class Cart:
    def __init__(self):
        self.products = []

    def add_product(self, product: Product):
        self.products.append(product)
        print(f"{product.name} sepete eklendi.")

    def calculate_total(self):
        total = sum(product.price for product in self.products)
        print(f"Toplam tutar: {total} TL")
```

```python
Kitap = Product("Kitap" , 30)
Kalem = Product("Kalem" , 5)

cart = Cart()
cart.add_product(Kitap)
cart.add_product(Kalem)
cart.calculate_total()
```

**Soru 3:** Bir öğrenci yönetim sistemi için Ogrenci ve Ders adlı iki sınıf oluşturun. Ogrenci sınıfı, öğrenci adı ve numarasını içermeli.
Ders sınıfı, ders adı ve öğrenci listesini tutmalı, ogrenci_ekle fonksiyonuyla öğrenci ekleyebilmeli.

Cevap:

```python
class Student:
    def __init__(self, name: str, number: int):
        self.name = name
        self.number = number
```

```python
class Course:
    def __init__(self, name: str):
        self.name = name
        self.students = []

    def add_student(self, student: Student):
        self.students.append(student)
        print(f"{student.name} {self.name} dersine eklendi.")

    def list_students(self):
        print(f"{self.name} dersindeki öğrenciler:")
        for student in self.students:
            print(f"- {student.name}, No: {student.number}")
```

```python
# Kullanım
student1 = Student("Ayşe", 101)
student2 = Student("Mehmet", 102)

math = Course("Matematik")
math.add_student(student1)
math.add_student(student2)
math.list_students()
```

**Soru 4:** Bir otopark yönetim sistemi geliştirin. Arac sınıfı, aracın plakası ve marka bilgilerini içermelidir. Otopark sınıfı, park edilen araçları tutmalı, arac_park_et fonksiyonuyla araç park edebilmeli ve arac_listesi fonksiyonuyla park edilen araçları listeleyebilmelidir.

Cevap:

```python
class Car:
    def __init__(self, plate: str, brand: str):
        self.plate = plate
        self.brand = brand
```

```python
class ParkingLot:
    def __init__(self):
        self.car_list = []

    def park_car(self, car: Car):
        self.car_list.append(car)
        print(f"{car.plate} plakalı {car.brand} park edildi.")

    def list_cars(self):
        print("Park edilen araçlar:")
        for car in self.car_list:
            print(f"- Plaka: {car.plate}, Marka: {car.brand}")
```

```python
car1 = Car("34ABC123", "Toyota")
car2 = Car("35XYZ789", "Honda")

park = ParkingLot()
park.park_car(car1)
park.park_car(car2)
park.list_cars()
```

**Soru 5:** Bir bankada hesap yönetim sistemi oluşturun. BankaHesabi sınıfı, hesap sahibinin adını ve bakiyesini tutmalı. para_yatir ve para_cek fonksiyonları ile hesap işlemlerini gerçekleştirin. Eğer bakiye yetersizse kullanıcıyı uyaran bir mesaj verin.

Cevap:

```python
class BankAccount:
    def __init__(self, account_holder_name: str, balance: float):
        self.account_holder_name = account_holder_name
        self.balance = balance

    def deposit(self, amount: float):
        self.balance += amount
        print(f"{amount} TL yatırıldı. Güncel bakiye: {self.balance} TL")

    def withdraw(self, amount: float):
        if self.balance >= amount:
            self.balance -= amount
            print(f"{amount} TL çekildi. Güncel bakiye: {self.balance} TL")
        else:
            print("Yetersiz bakiye. Mevcut bakiye:", self.balance)
```

```python
# Kullanım
hesap1 = BankAccount("Zeynep", 1000)
hesap1.deposit(200)
hesap1.withdraw(500)
hesap1.withdraw(1000)
```

**Soru 6:** Bir kütüphane yönetim sistemi geliştirin. Bu sistemde, kitapları yönetmek ve üyelerin ödünç alıp verme işlemlerini takip etmek için Kitap ve Kutuphane adında iki sınıf oluşturun.

Kitap sınıfı:
*   Kitabın adını ve mevcut stok miktarını tutmalıdır.
*   **stok_guncelle** adlı bir fonksiyonla stokları güncelleyebilmelidir.

Kutuphane sınıfı:
*   Birden fazla kitabı yönetmek için bir kitaplar listesi içermelidir.
*   **kitap_ekle** fonksiyonuyla kütüphaneye yeni bir kitap ekleyebilmelidir.
*   **odunc_al** fonksiyonuyla bir üyeye kitap ödünç vermelidir. Eğer kitap stokta yoksa, kullanıcıya uygun bir mesaj vermelidir.
*   **geri_ver** fonksiyonuyla ödünç alınan bir kitabı geri almalıdır.
*   **stok_goruntule** fonksiyonuyla kütüphanedeki kitapların mevcut stok durumunu listelemelidir.

Cevap:

```python
class Book:
    def __init__(self, book_name: str, stock: int):
        self.book_name = book_name
        self.stock = stock

    def update_stock(self, amount: int):
        self.stock += amount
```

```python

class Library:
    def __init__(self):
        self.books = []
    
    def add_book(self, book: Book):
        self.books.append(book)
        print(f"{book.book_name} kütüphaneye eklendi.")

    def borrow_book(self, book: Book):
        if book.stock > 0:
            book.stock -= 1
            print(f"{book.book_name} kitabı ödünç alındı. Kalan stok: {book.stock}")
        else:
            print(f"{book.book_name} kitabı şu anda stokta yok.")

    def return_book(self, book: Book):
        book.stock += 1
        print(f"{book.book_name} kitabı geri verildi. Güncel stok: {book.stock}")

    def view_stock(self):
        print("Kütüphanedeki kitapların stok durumu:")
        for book in self.books:
            print(f"- {book.book_name}: {book.stock} adet")
```

```python

# Create books
python_book = Book("Python Programlama", 5)
data_science_book = Book("Veri Bilimi", 3)
machine_learning_book = Book("Makine Öğrenimi", 0)

library = Library()

for book in (python_book,data_science_book,machine_learning_book): library.add_book(book)
```

```python
# Stok durumunu görüntüle
library.view_stock()

# Kitap ödünç alma işlemleri
library.borrow_book(python_book)                       # Stok: 4 olacak
library.borrow_book(data_science_book)                 # Stokta yok uyarısı verecek
library.borrow_book(machine_learning_book)             # Stok: 2 olacak

# Kitap geri verme işlemi
library.return_book(python_book)                    # Stok: 5 olacak

# Güncel stok durumunu tekrar görüntüle
library.view_stock()
```

**Soru 7:** Bir e-ticaret uygulaması için bir alışveriş sepeti yönetim sistemi geliştirin. Bu sistemde, ürünlerin ve sepetin yönetimini sağlayan iki ana sınıf olacak: Urun ve Sepet.

İstenen İşlevler:

Urun sınıfı:
*   Her ürünün bir adı, fiyatı ve stok miktarı olmalıdır.
*   Stok güncellemek için **stok_guncelle** adlı bir fonksiyon içermelidir.

Sepet sınıfı:

*   Bir urunler listesi içermelidir, bu listede kullanıcı tarafından sepete eklenen ürünler ve miktarları olmalıdır.
*   **urun_ekle** fonksiyonuyla sepete yeni bir ürün ekleyebilmelidir. Eklenmek istenen miktar stoktan fazla ise uygun bir mesaj vermelidir.
*   **urun_sil** fonksiyonuyla sepetten bir ürünü çıkarabilmelidir.
*   **toplam_fiyat** fonksiyonuyla sepetteki tüm ürünlerin toplam fiyatını hesaplayıp döndürmelidir.
*   **sepet_goruntule** fonksiyonuyla sepetteki ürünleri ve her birinin miktarını listelemelidir.

Çözüm:

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def update_stock(self, amount):
        self.stock += amount
```

```python

class Cart:
    def __init__(self):
        self.products = {}

    def add_product(self, product, amount):
        if product.stock >= amount:
            if product in self.products:
                self.products[product] += amount
            else:
                self.products[product] = amount
            product.update_stock(-amount)
            print(f"{amount} adet {product.name} sepete eklendi.")
        else:
            print(f"{product.name} stokta yeterli değil. Mevcut stok: {product.stock}")

    def remove_product(self, product, amount):
        if product in self.products:
            if self.products[product] >= amount:
                self.products[product] -= amount
                product.update_stock(amount)
                print(f"{amount} adet {product.name} sepetten çıkarıldı.")
            else:
                print(f"Sepette {product.name} ürününden {amount} adet bulunmamaktadır.")
        else:
            print(f"Sepette {product.name} ürününden hiç bulunmamaktadır.")

    def total_price(self):
        total = 0
        for product, amount in self.products.items():
            total += product.price * amount
        return total

    def display_cart(self):
        print("Sepetinizdeki ürünler:")
        for product, amount in self.products.items():
            print(f"- {product.name}: {amount} adet, Fiyat: {product.price} TL")
        print(f"Toplam Fiyat: {self.total_price()} TL")
```

```python
# Ürünleri oluşturalım
computer = Product("Bilgisayar", 15000, 10)
phone = Product("Telefon", 8000, 5)
headphones = Product("Kulaklık", 500, 20)

# Sepeti oluşturalım
cart = Cart()
```

```python
# Ürünleri sepete ekleyelim
cart.add_product(computer, 2)           # Bilgisayar eklenir
cart.add_product(phone, 1)              # Telefon eklenir
cart.add_product(headphones, 3)         # Kulaklık eklenir
cart.add_product(headphones, 20)         # Yetersiz stok uyarısı verir

# Sepeti görüntüleyelim
cart.display_cart()
```

```python
# Sepetten bir ürün çıkaralım
cart.remove_product(phone, 1)

# Güncel sepet durumunu görüntüleyelim
cart.display_cart()
```

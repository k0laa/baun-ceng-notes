**3. Nesne Oluşturma:**

Soru 3: Kitap adında bir sınıf oluşturunuz. Bu sınıfın her bir kitabın adı ve yazarını içermesi gerekmektedir. Ayrıca, kitap bilgilerini ekrana yazdıran bir metod tanımlayın. Ardından, en az iki kitap nesnesi oluşturun.

```python
class Book:
    def __init__(self, name : str, writer : str):
        self.name = name
        self.writer = writer

    def get_info(self):
        print(f"Kitap Adı : {self.name}, Yazar : {self.writer} ")

b1 = Book("1984", "George Orwell")
b2 = Book("Küçük Prens", "Antoine de Saint-Exupéry")

b1.get_info()
b2.get_info()

```

**1. Senaryo: Kütüphane Yönetim Sistemi**


Bir kütüphane yönetim sistemi geliştirmek istiyorsunuz. Her kitabın adı, yazarı ve mevcut adet sayısı tutulmalı. Kullanıcılar kitap ödünç alabilecek ve kitap iade edebilecekler. Her işlem sonucunda kitap adedi güncellenmeli. Eğer ödünç alınmak istenen kitap yoksa kullanıcıya bunun mümkün olmadığı söylenmeli.

```python
class Book:
    def __init__(self, name : str, writer : str):
        self.name = name
        self.writer = writer

    def get_info(self):
        print(f"Kitap Adı : {self.name}, Yazar : {self.writer} ")

b1 = Book("1984","George Orwell")
b2 = Book("Küçük Prens","Antoine de Saint-Exupéry")
```

```python
class Library:
    books = {}

    def add_book(self, book : Book, count : int):
        self.books[book.name] = count

    def barrow(self, book_name : str):
      if self.books[book_name] > 0:
         self.books[book_name] -= 1
         print("iyi günlerde kullanın. 1 hafta süreniz var.")
      else:
        print("kitap şuan mevcut değil.")

    def return_book(self, book : Book):
        self.books[book.name] += 1
        print("iadeniz için teşekkür ederiz.")

```

```python
lib = Library()

lib.add_book(b1,25)
lib.add_book(b2,1)

lib.barrow(b2.name) #kitap al
lib.barrow(b2.name) #kitap al hata vermeli

lib.return_book(b2) #kitap ver

lib.barrow(b2.name) #kitap al
lib.barrow(b2.name) #kitap al hata vermeli
```

**2. Senaryo: Otomobil Servis Sistemi**


Bir otomobil servis sistemi tasarlıyorsunuz. Her arabanın markası, modeli ve kilometresi olmalı. Kullanıcı arabasını servise getirdiğinde kilometresi güncellenmeli ve bu bilgiyi takip edebilmelisiniz. Ayrıca, servise giren arabaların kaydını tutmalı ve bu kayıtları listeleyebilmelisiniz. (2 sınıf oluşturunuz)

```python
class Car:
    def __init__(self, brand, model, km):
        self.brand = brand
        self.model = model
        self.km = km
```

```python
class Service:
    cars = {}

    def car_arrived(self, car : Car):
      print(f"{car.brand} {car.model} servise getirildi. Güncel kilometre = {car.km} km")
      self.save_car(car)

    def save_car(self, car):
       self.cars[car] = car.km
       print(f"{car.brand} {car.model} servise kaydedildi.\n")

    def print_cars(self):
      print("\nServise gelen arabalar : ")
      for car,km in self.cars.items():
        print(f"  {car.brand} {car.model} : {km}")
      print()
```

```python
c1 = Car("Toyota", "Corolla", 234000)
c2 = Car("Ford", "Focus", 7550)

s = Service()

s.car_arrived(c1)
s.car_arrived(c2)
s.print_cars()

c1.km = 350000
s.car_arrived(c1)
s.print_cars()

```

**3. Senaryo: Banka Hesap Yönetimi**

Bir bankada hesap yönetim sistemi oluşturuyorsunuz. Her kullanıcının bir hesabı olacak ve hesap bakiyesi tutulacak. Kullanıcı para yatırabilecek ve para çekebilecek. Eğer kullanıcı bakiyesinden fazla para çekmeye çalışırsa, sistem buna izin vermemeli ve kullanıcıyı uyarmalı.

```python
class Account:
    def __init__(self, name : str, money : float):
        self.name = name
        self.money = money
```

```python
class Bank:
    accounts = {}

    def deposit(self, account : Account, money : float):
      account.money += money
      self.accounts[account] = account.money
      print(f"Para yatırıldı. Yatırılan tutar : {money}, Güncel bakiye : {account.money}")


    def withdraw(self, account : Account, money : float):
      account.money -= money

      if(account.money > 0):
        self.accounts[account] = account.money
        print(f"Para çekildi. Çekilen tutar : {money}, Güncel bakiye : {account.money}")
      else:
        account.money += money
        print("Para çekilemedi. Yetersiz bakiye.")

```

```python
a1 = Account("enes",50)
a2 = Account("buğra",2000)

b = Bank()

b.withdraw(a1,100) # hata vermeli
b.deposit(a1,100) # yatır
b.withdraw(a1,100) # çek
```

**4. Senaryo: Online Alışveriş Sepeti**

Bir online alışveriş sistemi için bir sepet yönetim sistemi tasarlıyorsunuz. Kullanıcı, sepete ürün ekleyebilecek ve sepetindeki ürünleri listeleyebilecek. Her ürünün adı, fiyatı ve adedi tutulacak. Kullanıcı toplam tutarı görebilmeli. (2 sınıf oluşturunuz)

```python
class Product:
    def __init__(self, name : str, price : float):
      self.name = name
      self.price = price

telefon = Product("Telefon", 5000)
kulaklık = Product("Kulaklık", 300)
```

```python
class Customer:
    cart = {}
    def add(self, product : Product):
      if product in self.cart:
        self.cart[product] += 1
      else:
        self.cart[product] = 1
      print(f"{product.name} sepete eklendi.")

    def remove(self, product : Product):
      if self.cart[product] == 1:
        del self.cart[product]
      else:
        self.cart[product] -= 1
      print(f"{product.name} sepetten çıkarıldı.")

    def show_cart(self):
      print("\nSepetinizdeki ürünler : ")
      total = 0
      for i,b in self.cart.items():
        total += i.price * b
        print(f"- {i.name}, Adet: {b}, Fiyat: {i.price}")
      print(f"Toplam tutar : {total}\n")


```

```python
c = Customer()

c.add(telefon)
c.add(kulaklık)
c.add(kulaklık)

c.show_cart()

c.remove(kulaklık)

c.show_cart()
```
